```python
import pandas as pd

df = pd.read_csv("datos/estadisticas.txt",
                 sep="\t",
                 encoding="latin-1",
                 header=5)

# Ver las estaciones disponibles
estaciones = df.iloc[:, 0].dropna().unique()
for e in estaciones:
    print(e)
```

```python
# Filtrar solo los datos de Mar del Plata
mdp = df[df.iloc[:, 0] == "MAR DEL PLATA AERO"]

# Ver qué variables tiene
for _, fila in mdp.iterrows():
    print(fila.iloc[1])
```

```python
meses = ["Ene","Feb","Mar","Abr","May","Jun","Jul","Ago","Sep","Oct","Nov","Dic"]

# Filtrar Mar del Plata
mdp = df[df.iloc[:, 0] == "MAR DEL PLATA AERO"].copy()
mdp.columns = ["Estacion", "Variable"] + meses

# Ver los datos
print(mdp[["Variable"] + meses].to_string(index=False))
```
